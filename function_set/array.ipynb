```python
np.where(cond, x,y) is basically NumPy's 

```

```python
import numpy as np
a=np.array([1,2,3,4, 8, 10])
result = np.where(a%2==0, a, 0)
print(result)
```

```python
import numpy as np
a = np.array([1,5,8,2,10])
result = np.where(a>5, a,0)
print(result)
```

```python
import numpy as np
b = np.array([[1,2,4,6,8]])
result = np.where(b%3==0, "It's divisible by 3", "It's not divisible by 3")
print(result)
```
